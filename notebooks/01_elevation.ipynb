## Why this DEM
ALOS AW3D30 is a ~30 m *surface*. In forest it sits on the canopy. Slope is
smoothed: Extreme cliffs and isolated low-angle start zones are both harder
to pick out. We use it because it is consistent across western Canada, not
because it is the sharpest local model.
