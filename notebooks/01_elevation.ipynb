# 01 — Elevation

**Time:** 09:25–10:20

ALOS AW3D30 is a ~30 m *surface* (here ~21 m after projecting to Canada Atlas
Lambert). In forest it sits on the canopy. Slope is smoothed: Extreme cliffs
and isolated low-angle start zones are both harder to pick out.

Live work is the bundled clip. Downloading a DEM for a home range is a
projector demo — not hotel wifi.


```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path(".").resolve()))
from workshop import DATA, OUT, need, read_raster, write_raster, hillshade, show
import matplotlib.pyplot as plt
import numpy as np

dem_path = need(DATA / "01_elevation" / "inputs" / "alos_aw3d30_connaught.tif")
z, profile, transform, crs = read_raster(dem_path)
cell = abs(transform.a)
print(crs, z.shape, f"cell {cell:.2f} m", f"elev {np.nanmin(z):.0f}–{np.nanmax(z):.0f} m")
```

```python
# Slope in degrees from the 3x3 neighbourhood (same idea as the PRA step).
dy, dx = np.gradient(z, cell, cell)
slope = np.degrees(np.arctan(np.hypot(dx, dy)))
hs = hillshade(z, cell)

work = OUT / "workshop"
slope_path = write_raster(work / "slope.tif", slope, profile)
hs_path = write_raster(work / "hillshade.tif", hs, profile)
print("wrote", slope_path.name, hs_path.name)

fig, axs = plt.subplots(1, 3, figsize=(14, 4.5))
show(z, "Elevation (m)", "terrain", ax=axs[0])
show(hs, "Hillshade", "gray", ax=axs[1])
show(slope, "Slope (deg)", "magma", vmin=0, vmax=50, ax=axs[2])
plt.tight_layout()
```

Drag `outputs/connaught/workshop/slope.tif` into QGIS if you want. The PRA
